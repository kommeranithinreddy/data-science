```python
#Inspect the dataset

scores = np.array([
    [78, 85, 92, 67],
    [88, 76, np.nan, 81],
    [90, 84, 79, 93],
    [65, 72, 68, 75],
    [92, 89, 94, 96]
], dtype=float)

print(scores.shape)
print(scores.ndim)
print(scores.size)
print(scores.dtype)
```

```python
#Find the missing value

print(np.isnan(scores))
print()
print(np.isnan(scores).sum())
```

```python
#Handle missing data
#Calculate the average score for each subject
subject_means = np.nanmean(scores, axis = 0)
print(subject_means)
```

```python
#Replace the missing value
scores = np.where(
    np.isnan(scores), 
    subject_means, 
    scores)

print(scores)
```

```python
#Calculate each student's average score
student_avg = scores.mean(axis=1)
print(student_avg)
```

```python
#Calculate each subject's average score
subject_avg = scores.mean(axis = 0)
print(subject_avg)
```

```python
#Find all scores greater than or equal to 90
print(scores[scores >= 90])
```

```python
#Create Pass/Fail results
print(np.where(student_avg >= 75, "Pass", "Fail"))
```

```python
#Add the bonus to every student
bonus = np.array([2, 0, 3, 1])
print(bonus)
scores_bonus = scores + bonus
print(scores_bonus)
```

```python
#Calculate the percentiles
print(np.percentile(scores_bonus, 25))
print(np.percentile(scores_bonus, 50))
print(np.percentile(scores_bonus, 75))
```

```python
#Calculate each student's weighted score

weights = np.array([0.20, 0.30, 0.25, 0.25])
weighted_scores = scores_bonus @ weights
print(weighted_scores)
```

```python
#Find the student with the highest weighted score

print(np.argmax(weighted_scores)) #argmax() gives the zero-based index, not the human-readable student number.
print(np.max(weighted_scores))
```
