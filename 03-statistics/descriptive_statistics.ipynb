# Descriptive Statistics

## Block 1: Measures of Central Tendency

Central tendency tells us the central or typical value of a dataset.

### 1. Mean
- Mean = Sum of values / Number of values
- Represents the average.
- Sensitive to extreme values (outliers).

### 2. Median
- The middle value after sorting the data.
- For an even number of values, average the two middle values.
- Less affected by outliers.

### 3. Mode
- The most frequently occurring value.
- Useful for numerical and categorical data.

### Quick Revision
- Mean = Average
- Median = Middle value
- Mode = Most frequent value

```python
import numpy as np
import pandas as pd
```

```python
salaries = [20000, 25000, 30000, 30000, 35000, 50000, 200000]
print(f'Mean : {np.mean(salaries)}')
print(f'Median : {np.median(salaries)}')
print(f'Mode : {pd.Series(salaries).mode().tolist()}')
```

## Block 2: Measures of Variability

Measures of variability tell us how spread out the data values are.

### 1. Range
- Difference between the maximum and minimum values.
- Formula: Range = Maximum - Minimum
- Simple, but affected by outliers.

### 2. Variance
- Measures how far values spread from the mean.
- Calculated using squared differences from the mean.
- Larger variance generally means greater spread.

### 3. Standard Deviation (SD)
- Measures how far values typically spread from the mean.
- It is the square root of variance.
- Uses the same units as the original data.
- Higher SD means greater spread.

### 4. Interquartile Range (IQR)
- Measures the spread of the middle 50% of the data.
- Formula: IQR = Q3 - Q1
- Q1 = 25th percentile.
- Q3 = 75th percentile.
- Less affected by outliers than range or SD.

### Quick Revision
- Range = Maximum - Minimum
- Variance = Average squared deviation from the mean
- SD = Square root of variance
- IQR = Q3 - Q1

```python
A = [40, 41, 42, 43, 44]
B = [10, 25, 40, 55, 70]

r1 = np.max(A) - np.min(A)
r2 = np.max(B) - np.min(B)
print(r1, r2)
```

## Block 3: Position of Data

Position measures tell us where a value lies within a dataset.

### 1. Quartiles

* Divide sorted data into four parts.
* Q1 = 25th percentile.
* Q2 = 50th percentile (Median).
* Q3 = 75th percentile.

### 2. Percentiles

* Divide data into 100 positions.
* The 25th percentile means approximately 25% of values are at or below that value.
* The 50th percentile is the median.
* The 90th percentile means approximately 90% of values are at or below that value.

### 3. Five-Number Summary

Contains five values:

* Minimum
* Q1
* Median (Q2)
* Q3
* Maximum

### Quick Revision

* Quartiles → Divide data into four parts.
* Percentiles → Show the relative position of values.
* Q2 = Median.
* Five-number summary → Minimum, Q1, Median, Q3, Maximum.
* IQR = Q3 − Q1.


```python
data = [10, 20, 30, 40, 50, 60, 70, 80, 90]
df = pd.Series(data)
print(df.describe().round(2))
```

## Block 4: Distribution of Data

**Meaning:** Distribution describes how values are spread across a dataset and how frequently they occur.

### 1. Frequency Distribution

* Shows how often each value or category occurs.
* `value_counts()` counts occurrences in Pandas.

### 2. Histogram

* Shows the distribution of numerical data using bars.
* Helps identify concentration, spread, gaps, and skewness.

### 3. Common Distribution Shapes

* **Symmetric:** Both sides are approximately balanced.
* **Right-skewed:** A long tail extends to the right; often mean > median.
* **Left-skewed:** A long tail extends to the left; often mean < median.
* **Uniform:** Values occur at roughly similar frequencies across a range.

### 4. Why Distribution Matters

* Helps identify unusual values and patterns.
* Helps us understand data before building ML models.

### Quick Revision

* Frequency distribution → How often values occur.
* Histogram → Visualizes numerical distributions.
* Shape → Shows how data is distributed.
* Always investigate unusual patterns before making conclusions.


```python
A = [10, 20, 30, 40, 50, 60, 70]
B = [10, 20, 20, 20, 30, 40, 100]
C = [10, 60, 60, 60, 70, 80, 90]

print(np.mean(A), np.median(A))
print(np.mean(B), np.median(B))
print(np.mean(C), np.median(C))
```

## Block 5: Skewness and Outliers

### 1. Skewness

Skewness describes the asymmetry of a data distribution.

* **Symmetric:** Both sides are approximately balanced.
* **Right-skewed:** Long tail toward higher values; often mean > median.
* **Left-skewed:** Long tail toward lower values; often mean < median.

### 2. Outliers

* Outliers are values unusually far from most other observations.
* They can affect the mean, range, and standard deviation.
* Outliers are not always errors.
* Investigate outliers before deciding how to handle them.

### 3. IQR Method

Used to identify potential outliers.

* IQR = Q3 − Q1
* Lower bound = Q1 − 1.5 × IQR
* Upper bound = Q3 + 1.5 × IQR
* Values outside these bounds are potential outliers.

### Quick Revision

* Skewness → Asymmetry of a distribution.
* Outlier → Unusually distant observation.
* IQR method → Identifies potential outliers.
* Never delete an outlier without investigation.


```python
data = [10, 12, 13, 14, 15, 16, 18, 100]
print(np.mean(data), np.median(data))
```

## Block 6: Relationships Between Variables

Relationships describe how two variables change in relation to each other.

### 1. Covariance

* Indicates whether two variables tend to move together or in opposite directions.
* Positive covariance: They tend to move in the same direction.
* Negative covariance: They tend to move in opposite directions.
* Its magnitude depends on the variables' scales.

### 2. Correlation

* Measures the direction and strength of a linear relationship.
* Correlation coefficient (r) ranges from -1 to +1.
* r = +1: Perfect positive linear relationship.
* r = -1: Perfect negative linear relationship.
* r ≈ 0: Little or no linear relationship.

### 3. Correlation Does Not Imply Causation

* Correlation does not prove that one variable causes another.
* Other variables may influence both.

### 4. Python

* `df.corr(numeric_only=True)` calculates correlations between numerical columns.
* A scatter plot helps visualize relationships between two numerical variables.

### Quick Revision

* Positive correlation → Variables tend to increase together.
* Negative correlation → One tends to increase as the other decreases.
* Correlation near zero → Little linear relationship.
* Correlation ≠ Causation.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "experience": [1, 2, 3, 4, 5],
    "salary": [20000, 25000, 30000, 35000, 40000],
    "hours_sleep": [8, 7, 6, 8, 7]
})

# Calculate correlations between numerical columns
print(df.corr(numeric_only=True).round(2))

# Visualize experience versus salary
plt.scatter(df["experience"], df["salary"])
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Experience vs Salary")
plt.show()
```

## Block 7: Descriptive Statistics for Categorical Data

Categorical data represents groups or labels, such as department, gender, or product category.

### 1. Frequency Counts

* Count how often each category occurs.
* Pandas: `df["column"].value_counts()`

### 2. Proportions and Percentages

* Proportion = Category count / Total count
* Percentage = Proportion × 100
* Pandas: `value_counts(normalize=True) * 100`

### 3. Mode

* The most frequently occurring category.
* Pandas: `df["column"].mode()`

### 4. Grouped Summaries

* Summarize numerical values for each category.
* Pandas: `df.groupby("category")["numeric_column"].mean()`

### Quick Revision

* `value_counts()` → Counts each category.
* `normalize=True` → Returns proportions.
* Multiply by 100 → Converts proportions to percentages.
* `mode()` → Finds the most frequent category.
* `groupby()` → Summarizes data by category.


```python
df = pd.DataFrame({
    "department": [
        "IT", "HR", "IT", "Sales",
        "IT", "HR", "Sales", "IT"
    ],
    "salary": [
        50000, 30000, 60000, 40000,
        55000, 35000, 45000, 65000
    ]
})
print(df["department"].value_counts())
print(df["department"].value_counts(
    normalize = True
).round(2)*100)

print(df.groupby(
    "department"
)["salary"].mean())
```

# Block 8 — Descriptive Statistics in Python

## 1. Numerical Summary

* `mean()` — average value.
* `median()` — middle value.
* `mode()` — most frequent value.
* `min()` and `max()` — smallest and largest values.
* `std()` — measures how spread out values are.
* `quantile()` — calculates quartiles and other percentiles.
* `describe()` — provides a quick statistical summary.

## 2. Categorical Summary

* `value_counts()` — counts each category.
* `value_counts(normalize=True)` — calculates category proportions.
* `mode()` — returns the most frequent category.
* `groupby()` — summarizes numerical values for each category.

## 3. Data Quality Checks

* `isnull().sum()` — counts missing values in each column.
* `duplicated().sum()` — counts duplicate rows.
* `shape` — shows the number of rows and columns.
* `dtypes` — shows each column's data type.

## 4. Important Reminders

* The mean is sensitive to extreme values; the median is more resistant.
* Standard deviation measures spread.
* Quartiles and IQR help describe the spread of the middle 50% of data.
* Correlation describes a linear relationship, not proof of causation.
* Outliers should be investigated before being removed.
* Summary statistics describe the data; they do not automatically explain why a pattern exists.

```python
print(df.shape)
print(df.dtypes)
print("count of missing values")
print(df.isna().sum())
print("count of duplicated rows:")
print(df.duplicated().sum())
print("Mean salary:", df["salary"].mean())
print("Median salary:", df["salary"].median())
```
