````python
import numpy as np

employees = pd.DataFrame({
    "Name": ["Anil", "Rahul", "Priya", "Sneha", "Vikram", "Anil"],
    "Age": [24, 27, np.nan, 29, 31, 24],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "IT"],
    "Salary": [45000, 50000, 41000, 65000, 72000, 45000]
})
````

````python
employees["Salary_Lakhs"] = employees["Salary"] / 100000

employees["Age_Next_Year"] = employees["Age"] + 1

employees["Salary_Level"] = np.where(
    employees["Salary"] >= 60000,
    "High",
    "Normal"
)

employees.info()
print(employees["Salary_Lakhs"])
print(employees["Age_Next_Year"])
print(employees["Salary_Level"])
print(employees)
````

````python
print(employees)
print("-" * 50)
employees = employees.drop(
    columns = ["Salary_Lakhs", "Age_Next_Year"]
)
print(employees)
````

````python
print(employees.sort_values("Salary"))
print("-" * 50)
print(employees.sort_values("Salary", ascending = False))
print("-" * 50)
print(employees.sort_values(
    ["Department", "Salary"],
    ascending = [True, False]
))
````

````python
print(employees.duplicated())
print("-" * 50)
print(employees.duplicated().sum())
print("-" * 50)
print(employees[employees.duplicated()])
print("-" * 50)
print(employees)
print("-" * 50)
employees = employees.drop_duplicates()
print(employees)
````

````python
print(employees.isna())
print("-" * 50)
print(employees.isna().sum())
print("-" * 50)
print(employees[employees["Age"].isna()])
print("-" * 50)
print(employees["Age"])
print("-" * 50)
median_age = employees["Age"].median()
employees["Age"] = employees["Age"].fillna(median_age)
print(employees["Age"])
print("-" * 50)
print(employees.isna())
````

Using Pandas, answer:

> Which city generated the highest total sales?


````python
sales = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "City": [
        "Mumbai", "Delhi", "Mumbai", "Bangalore",
        "Delhi", "Mumbai", "Bangalore", "Delhi",
        "Mumbai", "Bangalore"
    ],
    "Category": [
        "Electronics", "Clothing", "Clothing", "Electronics",
        "Electronics", "Clothing", "Clothing", "Electronics",
        "Electronics", "Clothing"
    ],
    "Sales": [
        5000, 3000, 2500, 7000, 6500,
        4000, 3500, 8000, 6000, 4500
    ]
})
````

````python
print(sales["City"].value_counts())
print("*" * 50)
print(sales["Category"].value_counts())
print("*" * 50)
````

````python
print(sales.groupby(
    "City"
)["Sales"].mean().round(2))
print("*" * 50)

print(sales.groupby(
    "City"
)["Sales"].sum())
print("*" * 50)

print(sales.groupby(
    "City"
)["Sales"].max())
print("*" * 50)

print(sales.groupby(
    "City"
)["Sales"].min())
````

````python
print(sales.groupby(
    "Category"
)["Sales"].agg(
    ['min', 'max', 'mean', 'sum']
))
````

````python
print(sales.groupby(
    "City"
).agg(
    Average_Sales = ("Sales", "mean"),
    Total_Sales = ("Sales", "sum"),
    Maximum_Sales = ("Sales", "max")
))
````

````python
print(sales.groupby(
    ["City", "Category"]
)["Sales"].sum())
````

````python
print(sales[sales["Sales"] > 4000].groupby(
    "City"
)["Sales"].mean())
````

````python
print(sales.groupby(
    "Category"
)["Sales"].sum().sort_values(
    ascending = False
))
````

````python
print(sales.groupby(
    "City"
)["Sales"].sum().sort_values(
    ascending = False
))
#Both Delhi and Mumbai generated the highest total sales
````
